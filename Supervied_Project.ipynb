````python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import tensorflow.keras.datasets.mnist as mnsit
from sklearn.model_selection import train_test_split
from tensorflow.keras import layers, models, Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, Activation, LeakyReLU
from tensorflow.keras.activations import swish
from tensorflow.keras.optimizers import SGD, Adam, RMSprop
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import time
from tensorflow.keras.callbacks import LambdaCallback
from keras import callbacks
````

## **Load & Preprocess Data**

````python
mnist = keras.datasets.mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()

print('Train Images Shape:      ', X_train.shape)
print('Train Labels Shape:      ', y_train.shape)

print('\nTest Images Shape:       ', X_test.shape)
print('Test Labels Shape:       ', y_test.shape)

X_train = tf.keras.utils.normalize(X_train, axis=1)
X_test = tf.keras.utils.normalize(X_test, axis=1)

y_train_svm = y_train
y_test_svm = y_test
````

- Data has been normalized (each pixel  value divided by 255)
- no modifications yet in the shape of the input, since it will be different for each case. (some take the input flattened and some not)

# **Model(1) : SVM**

````python
x_train_reshaped = X_train.reshape(-1, 28 * 28).astype('float32')
x_test_reshaped = X_test.reshape(-1, 28 * 28).astype('float32')


svm_clf = SVC(kernel='rbf', C=5)

svm_clf.fit(x_train_reshaped, y_train_svm)

# Evaluate on test data
y_pred = svm_clf.predict(x_test_reshaped)
accuracy = accuracy_score(y_test_svm, y_pred)

print(f"SVM Test Accuracy: {accuracy:.4f}")
````

# **Model (2) : ANN**

````python
#Sequential -> left to right
model_2 = Sequential(name="model_2")

model_2.add(Flatten(input_shape=(28, 28))) #Input
model_2.add(Dense(128, activation='relu')) #Hidden layer 1
model_2.add(Dense(64, activation='relu')) #Hidder layer 2
model_2.add(Dense(10, activation='softmax')) #Output

model_2.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model_2.summary()

start_time = time.time()
model_2.fit(X_train, y_train
          ,epochs=15
          ,batch_size=64
          ,validation_split=0.1 # Sets 10% of training data for validation
          ,shuffle=True)
end_time = time.time()

total_training_time = end_time - start_time
avrg_per_epoch = total_training_time / 10

# Evaluate the model
start_test_time = time.time()
test_loss, test_acc = model_2.evaluate(X_test, y_test, verbose=1)
end_test_time = time.time()

total_test_time = end_test_time - start_test_time


print('\nTest Accuracy:', test_acc)
print('Test Loss:    ', test_loss)
print(f"\nTotal Training Time: {total_training_time:.2f} seconds")
print(f"\nAverage Training Time Per Epoch: {avrg_per_epoch:.2f} seconds")
print(f"\nTotal Test Time: {total_test_time:.2f} seconds")
````

# **Model(3) : Initial CNN**
- 1 con layer
- learning rate = 0.01
- Epochs = 1
- bathc_size = 32

````python
# adjusting input shape
print("X_train Before reshaping: ",X_train.shape)
X_train_conv = X_train.reshape(60000, 28, 28, 1)
X_test_conv = X_test.reshape(10000, 28, 28, 1)

print("X_train After reshaping: " ,X_train_conv.shape)
````

````python
# 1. Create the model
model_3  = Sequential(name='model_3')
# 2.Convolutional layer
model_3.add(Conv2D(32, (3, 3), input_shape=(28, 28, 1)))
# 3. Activation function
model_3.add(Activation('relu'))
# 4. Pooling
model_3.add(MaxPooling2D(pool_size=(2, 2)))
# 5. Flatten the image to 1D prepare it for the dense (Fully Connected) layer
model_3.add(Flatten())
model_3.add(Dense(128))
model_3.add(Activation('relu'))
# 6. Output Layer
model_3.add(Dense(10))
model_3.add(Activation('softmax'))

# Optimizer
sgd = SGD(momentum=0.9, learning_rate=0.01)
model_3.compile(optimizer=sgd, loss='sparse_categorical_crossentropy', metrics=['accuracy'])


# Time for training
start_train_time = time.time()

model_3.fit(X_train_conv, y_train,epochs=1 ,shuffle=True, batch_size=32, validation_split=0.1)

end_train_time = time.time()

total_train_time = end_train_time - start_train_time
avg_train_time_per_epoch = total_train_time / 1


print(f"Average Time per Epoch: {avg_train_time_per_epoch:.2f} seconds")

# Timer for testing
start_test_time = time.time()
test_loss, test_acc = model_3.evaluate(X_test_conv, y_test, verbose=0)
end_test_time = time.time()

total_test_time = end_test_time - start_test_time
print(f"Total Test Time: {total_test_time:.2f} seconds")
print('\nTest Accuracy: {:.2f}%'.format(test_acc * 100))
print('Test Loss:     {:.2f}%'.format(test_loss * 100))
````
